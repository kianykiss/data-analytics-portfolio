# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga?usp=sharing

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado

```python
# Vamos precisar instalar 
# pyautogui -> para automação do mouse e teclado
# pandas -> para manipulação e análise de dados
# openpyxl -> para ler arquivos excel

import pyautogui
import time 

# pyautogui.click -> clicar com o mouse
# pyautogui.write -> escrever um Texto
# pyautogui.press -> pressionar uma única tecla
# pyautogui.hotkey -> pressionar uma combinação de teclas  
# pyautogui.PAUSE = 1 # tempo de espera entre as ações (1 segundo) 
pyautogui.PAUSE = 1 # tempo de espera entre as ações (1 segundo)

# Passo a passo do desafio
# Passo 1: Entrar no sistema da empresa (link do drive)
pyautogui.press('win')
pyautogui.write('chrome') 
pyautogui.press('enter')

time.sleep(1) # esperar o navegador abrir
link = 'https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga'
pyautogui.write(link)
pyautogui.press('enter')

time.sleep(6) # esperar a página carregar
```

```python
# Passo 2: Navegar no sistema para encontrar a base de dados
pyautogui.click(x=403, y=460, clicks=2) # clicar na pasta "Vendas"
time.sleep(1) # esperar a pasta abrir


# Passo 3: Exportar a base de dados (baixar o arquivo)
pyautogui.click(x=414, y=455, clicks=1) # "selecionar arquivo" outra opção seria button='right'para clicar com botão direito e depois clicar em "baixar"
pyautogui.click(x=1791, y=455, clicks=1) # "clicar 3 pontinhos"
pyautogui.click(x=1589, y=568, clicks=1) # "clicar baixar"
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas as pd

# Passo 4: Calcular os indicadores (Faturamento e quantidade de produtos vendidos)


# abrir a base de dados com o pandas
caminho = r'C:\Users\kiany\Downloads\Vendas - Dez.xlsx' # o r vem de raw string para evitar problemas com as barras invertidas
tabela = pd.read_excel(caminho)

# ver as informações da base de dados
display(tabela) 

# somar o faturamento total
faturamento = tabela['Valor Final'].sum()

# somar a quantidade de produtos vendidos
qtde_produtos = tabela['Quantidade'].sum()

# exibir os indicadores
print(faturamento)
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
# Passo 5: Enviar as informações por e-mail
# pyautogui.press('win')
# pyautogui.write('chrome') 
# pyautogui.press('enter')
# time.sleep(2) # esperar o navegador abrir


import pyperclip


pyautogui.hotkey('ctrl', 't') # abrir nova aba
pyautogui.write('http://mail.google.com')
pyautogui.press('enter')
time.sleep(8) # esperar o gmail abrir

pyautogui.click(x=186, y=260, clicks=1) # clicar em "escrever"
time.sleep(6)# esperar o gmail abrir

# pra quem enviar o e-mail
pyautogui.write('kiany_guerra+codigo@hotmail.com')
pyautogui.press('tab') # selecionar e-mail
pyautogui.press('tab') # pular para o campo de assunto

# qual o assunto do e-mail
pyperclip.copy('Relatório de Vendas - Dezembro') # copiar o assunto do e-mail para a área de transferência
pyautogui.hotkey('ctrl', 'v') # colar o assunto do e-mail

pyautogui.press('tab') # pular para o campo corpo do e-mail

# qual o conteúdo do e-mail
texto = f'''
Prezados, 

Segue o relatório de vendas de hoje.

Faturamento: R${faturamento:,.2f}
Quantidade de produtos vendidos: {qtde_produtos:,}


Qualquer dúvida estou à disposição.
Att,
Kiany 

''' # as três aspas permitem escrever um texto em várias linhas

pyperclip.copy(texto)
pyautogui.hotkey('ctrl', 'v')

# enviar o e-mail
pyautogui.click(x=1219, y=972, clicks=1)
```
